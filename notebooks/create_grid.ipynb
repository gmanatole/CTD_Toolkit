```python
from ctd_toolkit.grid import *
```

```python
lat_range = SpatialRange(-90, -40, 1)
lon_range = SpatialRange(-180, 180, 1)
depth_range = DepthRange(0, 500, 10)
time_range = TemporalRange("2004-01-01", "2023-12-31", "1D")

grid = SpatioTemporalGrid(
    latitude=lat_range,
    longitude=lon_range,
    depth = depth_range,
    time=time_range,
    crs="EPSG:4326",
    chunks={"time": 10})

ds = grid.build(lazy=True)

print(ds)
print("Shape:", grid.shape)
```
